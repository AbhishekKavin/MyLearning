```python
from transformers import AutoTokenizer, AutoModel
import torch
import numpy as np
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")
```

```python
sentences = [
    "The cat sat on the mat.",
    "The dog slept on the floor.",
    "I love natural language processing."
]
```

# Embedding (CLS)

```python
def get_embeddings(sentences, model, tokenizer):
    encoded_input = tokenizer(sentences, padding = True, truncation = True, return_tensors = 'pt')
    with torch.no_grad():
        model_output = model(**encoded_input)

    sentence_embeddings = model_output.last_hidden_state[:, 0,:]

    return sentence_embeddings.numpy()
```

```python
embeddings = get_embeddings(sentences, model, tokenizer)
print(f"Embedding shape: {embeddings.shape}")
print(f"First 5 dimensions of the sentences' embeddings:\n{np.round(embeddings[:, :5], 3)}")
```

# Embedding (Mean Pooling)
Instead of relying solely on the <b>[CLS]</b> token embedding to represent a sentence, one can also use <b>mean pooling</b>, which <b>averages the embeddings of all output tokens</b>. This approach often yields a more balanced and comprehensive representation of the sentence.
