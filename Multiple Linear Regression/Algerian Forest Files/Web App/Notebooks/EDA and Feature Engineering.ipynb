# Algerian Forest Files
The Algerian Forest Fires Dataset is a well-known dataset used in machine learning and data science for predicting forest fire occurrences based on weather conditions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
dataset = pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv',header = 1)
```

```python
dataset.head()
```

```python
dataset.info()
```

## Data Cleaning


```python
## Missing values
dataset.isnull().sum()
```

```python
dataset[dataset.isnull().any(axis=1)]
```

```python
dataset.loc[:122,'Region'] = 0
dataset.loc[122:,'Region'] = 1
```

```python
dataset.info()
```

```python
dataset[['Region']] = dataset[['Region']].astype('int64')
```

```python
dataset.head()
```

```python
dataset[dataset.isnull().any(axis=1)]
```

```python
dataset.dropna(inplace=True)
```

```python
dataset.head()
```

```python
dataset.isnull().sum()
```

```python
dataset.reset_index(drop=True,inplace=True)
```

```python
dataset.iloc[[122]]
```

```python
dataset.drop(122,inplace=True)
```

```python
dataset.reset_index(drop=True,inplace=True)
```

```python
dataset.columns
```

```python
## Fix spaces in column names
dataset.columns = dataset.columns.str.strip()
```

```python
dataset.columns
```

```python
dataset.info()
```

## Change the required columns to appropriate data types

```python
dataset.head(5)
```

```python

dataset[['month', 'year', 'Temperature', 'RH', 'Ws']] = dataset[['month', 'year', 'Temperature', 'RH','Ws']].astype('int64')
```

### Changing the other columns to float


```python
objects = [features for features in dataset.columns if dataset[features].dtypes=='O']
```

```python
for i in objects:
    if i!='Classes':
        dataset[i] = dataset[i].astype(float)
```

```python
dataset.info()
```

```python
dataset.describe()
```

```python
## Saving the clean dataset
dataset.to_csv('Algerian_forest_fires_cleaned.csv',index=False)
```

# EDA

```python
# Drop unwanted features
df = dataset.drop(columns = ['day','month','year'])
```

```python
df.head()
```

```python
df['Classes'].value_counts()
```

```python
# Encode
df['Classes'] = np.where(df['Classes'].str.contains('not fire'),0,1)
```

```python
df.head()
```

```python
## Plot density plot for all features
plt.style.use('seaborn-v0_8')
df.hist(bins=50, figsize=(20,15))
plt.show()
```

```python
## Percentage for pie chart
perc = df['Classes'].value_counts(normalize=True)*100
```

```python
class_labels = ['Fire','Not Fire']
plt.figure(figsize=(12,7))
plt.pie(perc,labels = class_labels, autopct = '%1.1f%%')
plt.title('Distribution of Fire and Not Fire Classes', fontsize=16)
```

```python
## Correlation matrix 
df.corr()
```

```python
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
```

```python
sns.boxplot(df['FWI'],color = 'green')
```

```python
## Monthly Fire Analysis
dftemp = dataset.loc[dataset['Region']==1]
dftemp['Classes'] = np.where(dftemp['Classes'].str.contains('not fire'),"Not Fire","Fire")
plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',data=dftemp,hue='Classes',palette='Set1')
plt.ylabel('Number of Fires', weight = 'bold', fontsize=14)
plt.xlabel('Month', fontsize=14)
plt.title('Monthly Fire Analysis in Region 1', weight = 'bold',fontsize=16)
```

```python
## Monthly Fire Analysis
dftemp = dataset.loc[dataset['Region']==0]
dftemp['Classes'] = np.where(dftemp['Classes'].str.contains('not fire'),"Not Fire","Fire")
plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',data=dftemp,hue='Classes',palette='Set1')
plt.ylabel('Number of Fires', weight = 'bold', fontsize=14)
plt.xlabel('Month', fontsize=14)
plt.title('Monthly Fire Analysis in Region 0', weight = 'bold',fontsize=16)
```
