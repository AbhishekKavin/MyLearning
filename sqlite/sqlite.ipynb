# Importing the libraries

```python
import sqlite3
```

# Connecring to SQLite3

```python
connection = sqlite3.connect('example.db')
```

# Creating Cursor

```python
cursor = connection.cursor()
```

# Creating a table

```python
cursor.execute('''
Create Table If Not Exists employees(
            id Integer Primary Key,
            name Text Not Null,
            age Integer,
            department Text
            )
''')

## Commit the changes
connection.commit()
```

```python
cursor.execute('''
Select * from employees
''')
```

# Insert Data into the Table

```python

cursor.execute('''
Insert into employees(name,age,department)
               values('Krish',32,'Data Scientist')
''')

cursor.execute('''
Insert into employees(name,age,department)
               values('Abhi',28,'AI Expert')
''')

cursor.execute('''
Insert into employees(name,age,department)
               values('Ashi',26,'Fullstack Developer')
''')

cursor.execute('''
Insert into employees(name,age,department)
               values('Akshu',23,'AI Expert')
''')

cursor.execute('''
Insert into employees(name,age,department)
               values('Appa',62,'Finance Manager')
''')

### Commit the changes
connection.commit()
```

# Query the data from the table

```python

cursor.execute('''
Select * from employees
''')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

# Update the data

```python

cursor.execute('''
UPDATE employees
Set age = 34
where name = 'Krish'
''')

## Commit
connection.commit()
```

```python
cursor.execute('''
Select * from employees
''')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

# Delete the data

```python

cursor.execute('''
Delete from employees where name = 'Krish'
''')

connection.commit()
```

```python
cursor.execute('''
Select * from employees
''')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

# Project

```python
connection = sqlite3.connect('sales_data.db')
cursor = connection.cursor()


cursor.execute('''
CREATE TABLE IF NOT EXISTS sales(
            id INTEGER PRIMARY KEY,
            date TEXT NOT NULL,
            product TEXT NOT NULL,
            sales INTEGER,
            region TEXT
               )
''')

sales_data = [
    ('2023-01-02','Product1',100,'North'),
    ('2023-01-03','Product2',700,'South'),
    ('2023-01-04','Product3',600,'East'),
    ('2023-01-05','Product4',500,'West'),
    ('2023-01-06','Product5',400,'North'),
    ('2023-01-07','Product2',300,'South'),
    ('2023-01-08','Product3',200,'East'),
]

cursor.executemany('''
INSERT INTO sales(date,product,sales,region)
                   values(?,?,?,?)
''',sales_data)

connection.commit()
```

```python
cursor.execute('''
Select * from sales
''')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

# Close the connection

```python
connection.close()
```
