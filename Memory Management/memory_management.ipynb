# Python Memory Management

Python Memory Management is the process by which the Python interpreter allocates, uses, and deallocates computer memory for Python objects during program execution. It's an automated system that handles memory operations behind the scenes, freeing developers from manual memory management tasks.

### Reference Counting
Reference Counting is a simple, straightforward technique for automatic memory management. Its core idea is to keep track of the number of references, pointers, or handles to a resource (most commonly a block of memory or an object) and to deallocate that resource immediately when the number of references drops to zero.

It's a form of garbage collection, but unlike more complex tracing collectors (like Java's Garbage Collector), it does not need to pause the program to look for unreachable objects.

```python
import sys

a=[]
print(sys.getrefcount(a))
# One reference from variable 'a' and one from getrefcount function
```

```python
b = a
print(sys.getrefcount(b))
# Now 3 references: 'a', 'b', and getrefcount function
```

```python
del b
print(sys.getrefcount(a))
# Back to 2 references: 'a' and getrefcount function
```

### Garbage Collection
Garbage Collection is a form of automatic memory management. It is the process by which a programming language or runtime environment automatically reclaims memory that is no longer in use by the program. This "no longer used" memory is called garbage.

The primary goal of a garbage collector is to free programmers from the burden of manual memory management (e.g., using malloc/free in C or new/delete in C++), thereby preventing common and severe bugs like:
<ul>
<li>Memory Leaks: Forgetting to free memory, causing the program to slowly consume all available RAM.</li>

<li>Dangling Pointers: Using a pointer to memory that has already been freed.</li>

<li>Double Free Errors: Accidentally freeing the same block of memory twice.</li>
</ul>

```python
import gc
## enable garbage collector
gc.enable()
```

```python
gc.disable()
```

```python
gc.collect()  # Manually trigger garbage collection
```

```python
## Garbage collection stats
print(gc.get_stats())
```

```python
## get unreachable objects
print(gc.garbage)
```

```python
import gc

class MyObject:
    def __init__(self,name):
        self.name = name
        print(f"My Object {self.name} created")
    
    def __del__(self):
        print(f"My Object {self.name} destroyed")

## Create circular reference
obj1 = MyObject("obj1")
obj2 = MyObject("obj2")
obj1.ref = obj2
obj2.ref = obj1

del obj1
del obj2

## Manully trigger garbage collection
gc.collect()
```

### Generators for Memory Efficieny

Generators are a powerful tool in Python for creating memory-efficient programs by producing items one at a time on demand, rather than storing an entire sequence in memory at once. This technique, known as lazy evaluation, is ideal for working with large datasets or infinite streams of data.

```python
def generate_numbers(n):
    for i in range(n):
        yield i

for number in generate_numbers(100000):
    print(number)
    if number>10:
        break
```

```python
## Profiling Memory Usage with tracemalloc
import tracemalloc

def create_list():
    return [i for i in range(10000)]

def main():
    tracemalloc.start()

    create_list()

    snapshot = tracemalloc.take_snapshot()
    top_stats = snapshot.statistics('lineno')

    print("[ Top 10 ]")

    for stat in top_stats[:10]:
        print(stat)
```

```python
main()
```
